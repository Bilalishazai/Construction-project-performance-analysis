```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    "Activity": [
        "Excavation",
        "Foundation",
        "Columns",
        "Beams",
        "Brickwork"
    ],
    
    "Planned_Cost": [
        10000,
        25000,
        18000,
        20000,
        15000
    ],
    
    "Actual_Cost": [
        9500,
        27000,
        17000,
        22000,
        14000
    ],
    
    "Planned_Days": [
        5,
        10,
        8,
        10,
        12
    ],
    
    "Actual_Days": [
        6,
        12,
        7,
        11,
        10
    ]
}
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
df["Cost_Variance"] = df["Planned_Cost"] - df["Actual_Cost"]

print(df)
```

```python
print(df[["Activity", "Planned_Cost", "Actual_Cost", "Cost_Variance"]])
```

```python
df["Schedule_Variance"] = df["Planned_Days"] - df["Actual_Days"]

print(df[[
    "Activity",
    "Planned_Days",
    "Actual_Days",
    "Schedule_Variance"
]])
```

```python
plt.bar(df["Activity"], df["Schedule_Variance"])

plt.title("Schedule Variance by Activity")
plt.xlabel("Activity")
plt.ylabel("Variance (Days)")

plt.show()
```

```python
total_planned_cost = df["Planned_Cost"].sum()
total_actual_cost = df["Actual_Cost"].sum()

print("Total Planned Cost:", total_planned_cost)
print("Total Actual Cost:", total_actual_cost)
```

```python
total_cost_variance = total_planned_cost - total_actual_cost

print("Total Cost Variance:", total_cost_variance)
```

```python
total_planned_days = df["Planned_Days"].sum()
total_actual_days = df["Actual_Days"].sum()

print("Total Planned Days:", total_planned_days)
print("Total Actual Days:", total_actual_days)
```

```python
total_schedule_variance = total_planned_days - total_actual_days

print("Total Schedule Variance:", total_schedule_variance)
```

```python
plt.bar(df["Activity"], df["Planned_Cost"], label="Planned")
plt.bar(df["Activity"], df["Actual_Cost"], label="Actual")

plt.title("Planned Cost vs Actual Cost")
plt.xlabel("Activity")
plt.ylabel("Cost")

plt.legend()
plt.show()
```

```python
df["Cost_Variance_Percent"] = (
    df["Cost_Variance"] / df["Planned_Cost"]
) * 100

print(df[["Activity", "Cost_Variance_Percent"]])
```

```python
plt.bar(df["Activity"], df["Cost_Variance"])

plt.title("Cost Variance by Activity")
plt.xlabel("Activity")
plt.ylabel("Cost Variance")

plt.show()
```
